# SDXL Turbo Studio (Ultra-Lightweight)
Uses SDXL Turbo to avoid CUDA OOM errors. Fast and efficient.

```python
!pip install -q diffusers transformers accelerate torch torchvision gradio

import torch
from diffusers import AutoPipelineForText2Image
import gradio as gr

print("Loading SDXL Turbo (Ultra-Lightweight)...")
pipe = AutoPipelineForText2Image.from_pretrained(
    "stabilityai/sdxl-turbo",
    torch_dtype=torch.float16,
    variant="fp16"
).to("cuda")

def generate(prompt):
    return pipe(prompt=prompt, num_inference_steps=1, guidance_scale=0.0).images[0]

demo = gr.Interface(fn=generate, inputs=gr.Textbox(label="Prompt"), outputs=gr.Image(label="Output"), title="SDXL Turbo Studio")
demo.launch(share=True, debug=True)
```
